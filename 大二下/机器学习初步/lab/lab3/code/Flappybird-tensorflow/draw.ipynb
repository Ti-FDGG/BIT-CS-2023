```python
import pandas as pd
import matplotlib.pyplot as plt

df_record = pd.read_csv('record.csv')
df_record_aTN = pd.read_csv('record_addTargetNetwork.csv')
```

```python
last = 6300
```

```python
plt.figure(figsize=(10, 6))
plt.plot(range(1, df_record['r_sum'][:last].shape[0]+1), df_record['r_sum'][:last])
plt.xlabel('Episode')
plt.ylabel('Reward')
plt.title('DQN without Target Network')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(df_record_aTN['episode'][:last], df_record_aTN['r_sum'][:last])
plt.xlabel('Episode')
plt.ylabel('Reward')
plt.title('DQN with Target Network')
plt.show()
```

```python
plt.figure(figsize=(10, 6))

# Plot for DQN with Target Network
plt.plot(df_record_aTN['episode'][:last], df_record_aTN['r_sum'][:last], label='With Target Network')

# Plot for DQN without Target Network
plt.plot(range(1, df_record['r_sum'][:last].shape[0]+1), df_record['r_sum'][:last], label='Without Target Network')

# Add labels, title, and legend
plt.xlabel('Episode')
plt.ylabel('Reward')
plt.title('Without Target Network vs With Target Network')
plt.legend()

plt.show()
```
